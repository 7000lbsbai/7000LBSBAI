# Quarterly Sales Analysis

This notebook is a first look at how **Python code**, **notes**, and **chart output** all live together in one document.

We'll:
1. Create a small sales dataset with `pandas`
2. Look at the raw numbers in a table
3. Visualize trends with `matplotlib`

Run each cell from top to bottom with **Shift + Enter**.

```python
import pandas as pd
import matplotlib.pyplot as plt

# A small made-up sales dataset for a business analytics example
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Revenue_K": [42, 48, 45, 61, 58, 67],
    "New_Customers": [28, 31, 26, 39, 35, 44],
}
df = pd.DataFrame(data)
df
```

## Visualizing Monthly Revenue

A quick summary statistic first, then a bar chart makes the seasonal pattern easy to spot.

```python
print("Average monthly revenue: $%.1fK" % df["Revenue_K"].mean())
print("Best month:", df.loc[df["Revenue_K"].idxmax(), "Month"])
```

```python
plt.figure(figsize=(6, 4))
plt.bar(df["Month"], df["Revenue_K"], color="#4285F4")
plt.title("Monthly Revenue ($K)")
plt.ylabel("Revenue ($K)")
plt.show()
```

## New Customers Over Time

Line charts are useful for comparing a trend across categories — here, new customers acquired each month.

```python
plt.figure(figsize=(6, 4))
plt.plot(df["Month"], df["New_Customers"], marker="o", color="#EA4335")
plt.title("New Customers by Month")
plt.xlabel("Month")
plt.ylabel("New Customers")
plt.show()
```

## Try it yourself

Change the numbers in the `data` dictionary above and re-run the notebook to see the table and charts update automatically. This is the core idea behind notebooks: **code, output, and explanation stay together, in order, in one file.**
