## 🔴 Extension 4 — A glimpse of SciPy

**SciPy** is another package, with more specialist statistics. A **z-score** says how many standard deviations a value is from the mean; values more than 3 away are often treated as unusual. You will meet these ideas again in Week 5.

```python
# 🔴 Extension 4: z-scores with SciPy
from scipy import stats

df['z_score'] = stats.zscore(df['total_spend'])
df[df['z_score'].abs() > 3][['customer_id', 'product_name', 'total_spend', 'z_score']]
```
