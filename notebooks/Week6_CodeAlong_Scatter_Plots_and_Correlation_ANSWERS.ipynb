## Describing a relationship

Last week you described **one** variable. This week you look at **two variables together**. A good interpretation of a relationship covers:

| | Ask yourself |
|---|---|
| **Direction** | Positive (both rise together) or negative (one rises as the other falls)? |
| **Strength** | How closely do the points follow a line? What is the correlation (*r*)? |
| **Form** | A straight line, a curve, or no pattern? |
| **Outliers** | Are there unusual points that could change the result? |
| **So what?** | What does it mean for the business? |

### A rough guide to the correlation coefficient (*r*)

| *r* (positive or negative) | Strength |
|---|---|
| 0.7 to 1.0 | Strong |
| 0.4 to 0.7 | Moderate |
| 0.1 to 0.4 | Weak |
| Close to 0 | Little or no straight-line relationship |

> 📝 This is a rough guide: your statistics module may use slightly different cut-offs. *r* is always between −1 and 1.

## Set up: run this cell

````python
# ✍️ Set up: just run this cell
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

url = "https://raw.githubusercontent.com/7000lbsbai/7000LBSBAI/main/datasets/beanhouse_daily_weather.csv"
df = pd.read_csv(url)
print(f"Data loaded: {df.shape[0]} rows")
````

---
# Block 1: Two variables at once
*about 15 minutes*

A **scatter plot** shows two numeric variables together: one point for each row (here, each branch on each day).

The **correlation coefficient** (*r*) puts a number on how closely the points follow a straight line. In pandas, `.corr()` calculates it for two columns.

````python
# 👀 Block 1: a scatter plot of temperature and cold drinks, and the correlation
# Watch the demo first, then run this cell yourself

plt.figure(figsize=(10, 5))

sns.scatterplot(data=df, x='temperature_c', y='cold_drinks', alpha=0.6)

plt.title('Cold drinks sold and temperature')
plt.xlabel('Temperature (°C)')
plt.ylabel('Cold drinks sold per day')
plt.show()

r = df['temperature_c'].corr(df['cold_drinks'])
print(f"Correlation (r): {r:.2f}")
````

> 📝 `alpha=0.6` makes the points slightly see-through, so you can see where they overlap.

### ➡️ Adapt: temperature and hot drinks

Create the same chart for **temperature** and **hot drinks**, with a title and axis labels. Then print the correlation.

> 📝 **Hint:** copy the Watch cell, then change `'cold_drinks'` to `'hot_drinks'` and update the title and labels.

````python
# ➡️ Adapt: temperature and hot drinks
plt.figure(figsize=(10, 5))

sns.scatterplot(data=df, x='temperature_c', y='hot_drinks', alpha=0.6)

plt.title('Hot drinks sold and temperature')
plt.xlabel('Temperature (°C)')
plt.ylabel('Hot drinks sold per day')
plt.show()

r = df['temperature_c'].corr(df['hot_drinks'])
print(f"Correlation (r): {r:.2f}")

# r = -0.45: a moderate negative relationship
````

### 🔴 Go further: correlation with numpy

# Block 2: Many variables at once
*about 15 minutes*

With several numeric columns, a **correlation matrix** shows the correlation between **every pair** at once. A **heatmap** colours it, so the strongest relationships stand out:

- **Red:** positive · **Blue:** negative · **Pale:** close to 0
- `annot=True` writes the value of *r* in each square
- `vmin=-1, vmax=1` keeps the colour scale from −1 to 1

````python
# 👀 Block 2: a correlation heatmap
# Watch the demo first, then run this cell yourself

cols = ['revenue', 'customers', 'temperature_c', 'rainfall_mm', 'hot_drinks', 'cold_drinks', 'promo_spend']
corr = df[cols].corr().round(2)

plt.figure(figsize=(10, 7))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlations between the BeanHouse variables')
plt.tight_layout()
plt.show()
````

> 📝 Each variable has a correlation of 1 with **itself**, which is why the diagonal is dark red. The top-right and bottom-left halves are mirror images, so you only need to read one half.

### ➡️ Adapt: what is most related to revenue?

Use the `corr` table from the Watch cell to list how strongly **each variable** is related to **revenue**, from highest to lowest. Remove revenue itself from the list.

> 📝 **Hint:** select the `'revenue'` column of `corr`, then use `.drop('revenue')` to remove it, and `.sort_values(ascending=False)`.

````python
# ➡️ Adapt: what is most related to revenue?
corr['revenue'].drop('revenue').sort_values(ascending=False)

# customers 0.91, hot_drinks 0.72, cold_drinks 0.51, promo_spend 0.45,
# temperature_c 0.07, rainfall_mm -0.01
````

**✏️ MODEL ANSWER:**

A heatmap is a quick way to scan **every pair** at once and decide which relationships to look at more closely with scatter plots.

# Block 3: Groups and trend lines
*about 15 minutes*

`sns.regplot()` draws a scatter plot **with a trend line** (the line of best fit), so the direction is easier to see.

In Week 5 you found **hidden groups** in a distribution. Groups can also hide inside a relationship, so this cell draws the same data twice: once all together, and once coloured by branch with `hue='branch'`.

````python
# 👀 Block 3: promotion spend and revenue, all together and by branch
# Watch the demo first, then run this cell yourself

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Left: all branches together, with a trend line
sns.regplot(data=df, x='promo_spend', y='revenue', scatter_kws={'alpha': 0.5}, ax=axes[0])
axes[0].set_title('All branches together')
axes[0].set_xlabel('Promotion spend (£)')
axes[0].set_ylabel('Revenue (£)')

# Right: the same points, coloured by branch
sns.scatterplot(data=df, x='promo_spend', y='revenue', hue='branch', alpha=0.7, ax=axes[1])
axes[1].set_title('Coloured by branch')
axes[1].set_xlabel('Promotion spend (£)')
axes[1].set_ylabel('Revenue (£)')

plt.tight_layout()
plt.show()

print(f"Correlation, all branches: {df['promo_spend'].corr(df['revenue']):.2f}")
````

# Block 4: The limits of correlation
*about 15 minutes*

Correlation is useful, but it has limits. This cell shows two of them:

- **Left:** City Centre revenue and temperature. Look for the festival day.
- **Right:** a **made-up example** of customers at an outdoor kiosk, where trade is best on mild days and falls on very cold **and** very hot days.

````python
# 👀 Block 4: an outlier, and a curved relationship
# Watch the demo first, then run this cell yourself

city = df[df['branch'] == 'City Centre']

# A made-up example: customers at an outdoor kiosk peak on mild days
rng = np.random.default_rng(1)
kiosk_temp = np.linspace(0, 30, 60)
kiosk_customers = 200 - 0.8 * (kiosk_temp - 15) ** 2 + rng.normal(0, 12, 60)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.regplot(data=city, x='temperature_c', y='revenue', scatter_kws={'alpha': 0.6}, ax=axes[0])
axes[0].set_title('City Centre: revenue and temperature')
axes[0].set_xlabel('Temperature (°C)')
axes[0].set_ylabel('Revenue (£)')

sns.scatterplot(x=kiosk_temp, y=kiosk_customers, ax=axes[1])
axes[1].set_title('Made-up example: customers at an outdoor kiosk')
axes[1].set_xlabel('Temperature (°C)')
axes[1].set_ylabel('Customers')

plt.tight_layout()
plt.show()

print(f"City Centre, revenue and temperature: r = {city['temperature_c'].corr(city['revenue']):.2f}")
print(f"Kiosk, customers and temperature:     r = {np.corrcoef(kiosk_temp, kiosk_customers)[0, 1]:.2f}")
````

Remove the festival day from the City Centre data (keep revenue **below £1,485**, as in Week 5), then redraw the left-hand chart and print the correlation again.

### 💬 What does it mean?

- How much did one day change the correlation? What does that tell you?
- The kiosk example has r close to 0. Does that mean temperature has no effect on its customers?
- Temperature and cold drinks are strongly correlated (Block 1). Does that prove warm weather causes people to buy cold drinks?

**✏️ MODEL ANSWER:**

Removing the festival day makes the City Centre correlation fall from **0.23 to 0.01**: **one unusual day** created the appearance of a link between warm weather and revenue. Always look at the scatter plot for outliers before trusting *r*. In the kiosk example *r* is close to 0, but there is a **clear curved pattern**: *r* only measures **straight-line** relationships, so a value near 0 does not mean there is no relationship. Finally, **correlation is not causation**. Warm weather causing cold drink sales is very plausible, but the correlation alone does not prove it, and Block 3 showed how a third variable (the branch) can create a correlation with no cause behind it.

### 🔴 Go further: is the correlation significant?

Your statistics module covers significance testing. SciPy's `stats.pearsonr()` gives the correlation **and** its p-value. Use it for temperature and cold drinks.

````python
# 🔴 Go further: is the correlation significant?
from scipy import stats

r, p = stats.pearsonr(df['temperature_c'], df['cold_drinks'])
print(f"r = {r:.2f}, p-value = {p:.3g}")

# p is far below 0.05: very strong evidence of a relationship
# (but still not proof of cause)
````
