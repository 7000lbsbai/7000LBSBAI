## Step 4 — 🔧 Deal with missing values

There are two choices for each missing value:

| Choice | Code | When to use it |
|--------|------|---------------|
| **Fill** it in | `.fillna(value)` | When a sensible value exists, e.g. `'Unknown'` for text |
| **Drop** the row | `.dropna(subset=['column'])` | When the row is no use without that value |

- A missing **payment method** does not stop it being a sale, so we **fill** it with `'Unknown'`.
- A missing **quantity** means we cannot work out the order total, so we **drop** those rows.

## ✅ End of Code-along — Recap

You have cleaned a messy dataset using the **Spot → Fix → Check** workflow:

| Problem | Spot it with | Fix it with |
|---------|-------------|-------------|
| Duplicate rows | `df.duplicated().sum()` | `df = df.drop_duplicates()` |
| Missing text | `df.isna().sum()` | `df['col'] = df['col'].fillna('Unknown')` |
| Missing numbers | `df.isna().sum()` | `df = df.dropna(subset=['col'])` or `.fillna(df['col'].median())` |
| Numbers stored as text | `df.info()` | `.str.replace('£', '')` then `pd.to_numeric()` |
| Messy text | `df['col'].value_counts()` | `df['col'].str.strip().str.title()` |
| Impossible values | `df.describe()` | `df = df[df['col'] > 0]` |

⚠️ **Remember the golden rule:** save each fix back into `df` (or the column).
