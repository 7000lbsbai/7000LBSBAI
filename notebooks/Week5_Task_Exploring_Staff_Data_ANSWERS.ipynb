## Section 1: Load and check the data

````python
# ➡️ Task 1.1: import pandas, NumPy, matplotlib and seaborn, set the seaborn theme, then load the data
url = "https://raw.githubusercontent.com/7000lbsbai/7000LBSBAI/main/datasets/retailco_staff_clean.csv"

# Write your code below this line
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv(url)
print(f"Data loaded: {df.shape[0]} rows")
````

````python
# ➡️ Task 1.2: check the data types and that there are no missing values
df.info()

# ✅ 108 rows, no missing values; salary is int64, hours_per_week is float64
````

---
## Section 2: The salary distribution

Create a histogram of `salary` with a density curve, mark the **mean** and **median**, add a title and axis labels, and print the skew.

> 📝 **Hint:** code-along Block 1.

````python
# ➡️ Task 2.1: the salary distribution
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='salary', bins=20, kde=True)
plt.axvline(df['salary'].mean(), color='red', linestyle='--', label=f"Mean £{df['salary'].mean():,.0f}")
plt.axvline(df['salary'].median(), color='black', label=f"Median £{df['salary'].median():,.0f}")
plt.title('Distribution of annual salaries')
plt.xlabel('Salary (£)')
plt.ylabel('Number of staff')
plt.legend()
plt.show()

print(f"Skew: {df['salary'].skew():.2f}")

# ✅ Skew 0.88; mean £30,002, median £28,750
````

### 💬 What does it mean?

- What shape is the salary distribution?
- Where are the mean and median, and why do they differ?
- Which figure would you give the HR director as a typical salary?

**✏️ MODEL ANSWER:**

Salaries are **moderately right-skewed** (skew 0.88). Most staff earn between about £26,000 and £33,000, with a tail of higher salaries up to £47,750. The mean (£30,002) is above the median (£28,750) because the higher salaries pull it up, so the **median** is the better figure for a typical salary. The tail is worth looking at next: which departments do the higher salaries come from?

---
## Section 3: Salaries by department

Create a **boxplot** of salary for each department.

To make it easier to read, put the departments **in order of median salary**, highest first. A boxplot takes an `order=` argument, which is a list of the categories in the order you want. You can make that list like this:

```python
dept_order = df.groupby('department')['salary'].median().sort_values(ascending=False).index
```

Then print the median salary for each department, highest first.

> 📝 **Hint:** code-along Block 2 (the boxplot). Make the figure wide enough for six department names.

````python
# ➡️ Task 3.1: salary by department, in order of median salary
dept_order = df.groupby('department')['salary'].median().sort_values(ascending=False).index

plt.figure(figsize=(12, 5))
sns.boxplot(data=df, x='department', y='salary', order=dept_order)
plt.title('Annual salary by department (highest median first)')
plt.xlabel('')
plt.ylabel('Salary (£)')
plt.show()

print(df.groupby('department')['salary'].median().sort_values(ascending=False))

# ✅ Finance £37,500, Marketing £35,125, Sales £29,750, Operations £29,375,
#    Logistics £25,250, Customer Service £24,500
````

### 💬 What does it mean?

- Which departments are paid the most and the least?
- Which departments have the widest and narrowest spread of salaries?
- How does this help explain the shape you found in Section 2?

**✏️ MODEL ANSWER:**

**Finance** (median £37,500) and **Marketing** (£35,125) are paid the most; **Customer Service** (£24,500) and **Logistics** (£25,250) the least. Finance and Marketing also have the **widest** boxes, so pay varies a lot within them, while Customer Service has a very **narrow** box: salaries there are very similar. This explains the right skew in Section 2: most staff work in lower-paid departments, and the long tail of higher salaries comes mainly from Finance and Marketing — so the overall average hides big differences between departments.

## Section 4: Categorical distributions

`hours_per_week` only takes a few different values, so it behaves like a **category**. A count plot shows its distribution better than a histogram.

> 📝 **Hint:** code-along Block 4 (the count plot). `value_counts(normalize=True) * 100` gives percentages (Week 4).

````python
# ➡️ Task 4.1: a count plot of hours_per_week, then the percentage of staff on each number of hours
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='hours_per_week', color='steelblue')
plt.title('Contracted hours per week')
plt.xlabel('Hours per week')
plt.ylabel('Number of staff')
plt.show()

print((df['hours_per_week'].value_counts(normalize=True) * 100).round(1))

# ✅ 37.5 hours: 71.3% (77 staff); 22.5: 10.2%; 30 and 16: 9.3% each
````

````python
# ➡️ Task 4.2: a count plot of performance_rating
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='performance_rating', color='steelblue')
plt.title('Performance ratings (1 = lowest, 5 = highest)')
plt.xlabel('Performance rating')
plt.ylabel('Number of staff')
plt.show()

# ✅ Most common rating 3 (44 staff); only 2 staff rated 1 and 13 rated 5
````

### 💬 What does it mean?

- What do the two count plots tell you about working patterns and performance?
- What shape is the distribution of performance ratings?

**✏️ MODEL ANSWER:**

Most staff (**71%**) work full-time, 37.5 hours a week; the rest are spread fairly evenly across three part-time contracts (30, 22.5 and 16 hours). Performance ratings are **roughly symmetric** around the middle rating of **3** (44 staff), with slightly more staff rated 4 (30) than 2 (19), and very few at the extremes. This could suggest ratings cluster around the middle of the scale, so HR may want to check whether the scale separates strong and weak performance well.

## Section 5: London and the other offices

London salaries are often higher. Is that true at RetailCo?

1. Use `np.where()` to create a column called `location`: `'London'` if `office` is London, and `'Other'` otherwise.
2. Create **two charts side by side**: a boxplot of salary by `location`, and overlapping histograms of salary with `hue='location'`.
3. Print the median salary for each location.

> 📝 **Hint:** `np.where()` from Week 4; the side-by-side charts are code-along Block 2.

````python
# ➡️ Task 5.1: London compared with the other offices
df['location'] = np.where(df['office'] == 'London', 'London', 'Other')

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.boxplot(data=df, x='location', y='salary', ax=axes[0])
axes[0].set_title('Annual salary: London and other offices')
axes[0].set_xlabel('')
axes[0].set_ylabel('Salary (£)')

sns.histplot(data=df, x='salary', hue='location', bins=20, element='step', ax=axes[1])
axes[1].set_title('Salary distribution by location')
axes[1].set_xlabel('Salary (£)')
axes[1].set_ylabel('Number of staff')

plt.tight_layout()
plt.show()

print(df.groupby('location')['salary'].median())

# ✅ London median £30,125 (26 staff); other offices £28,125 (82 staff)
````

### 💬 What does it mean?

- How do London salaries compare with the other offices?
- Is the difference large or small compared with the differences between departments?
- The 'Other' group includes 7 staff whose office is Unknown. Why might that matter?

**✏️ MODEL ANSWER:**

London's median salary (**£30,125**) is about **£2,000 higher** than the other offices (£28,125), and its whole box sits a little higher. But this difference is **much smaller** than the gap between departments (over £13,000 between Finance and Customer Service), so department matters far more than location. The 'Other' group includes 7 staff whose office is Unknown: if some of them actually work in London, the comparison could change slightly, so this is worth noting as a limitation.

## Section 6: Choose the chart

This time **you** decide which chart to use. For each question, choose a suitable chart, create it, and explain your choice in one sentence in the interpretation cell.

> 📝 **Which chart?** One numeric variable → histogram. A numeric variable compared across groups → boxplot. How many in each category → count plot.

### ➡️ Task 6.1 — Which office has the most staff?

````python
# ➡️ Task 6.1: which office has the most staff?
office_order = df['office'].value_counts().index

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='office', order=office_order, color='steelblue')
plt.title('Number of staff in each office')
plt.xlabel('')
plt.ylabel('Number of staff')
plt.show()

# ✅ A count plot: London 26, Manchester 22, Glasgow 19, Leeds 18, Birmingham 16, Unknown 7
````

### ➡️ Task 6.2 — Do full-time staff earn more than part-time staff?

First, use `np.where()` to create a `contract` column: `'Full-time'` when `hours_per_week` is 35 or more, and `'Part-time'` otherwise. Then choose a chart to answer the question, and print the median salary for each contract type.

````python
# ➡️ Task 6.2: do full-time staff earn more than part-time staff?
df['contract'] = np.where(df['hours_per_week'] >= 35, 'Full-time', 'Part-time')

plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x='contract', y='salary')
plt.title('Annual salary: full-time and part-time staff')
plt.xlabel('')
plt.ylabel('Salary (£)')
plt.show()

print(df.groupby('contract')['salary'].median())

# ✅ A boxplot: full-time median £29,250 (77 staff), part-time £27,500 (31 staff)
````

### 💬 What does it mean?

- Which chart did you choose for each question, and why?
- Do full-time staff earn more than part-time staff? Is anything surprising about the result?

**✏️ MODEL ANSWER:**

For 6.1 I used a **count plot**, because the question asks how many staff are in each category (office). For 6.2 I used a **boxplot**, because it compares a numeric variable (salary) across groups (contract type). Full-time staff have a slightly higher median salary (**£29,250**) than part-time staff (**£27,500**), but the difference is surprisingly **small** — part-time staff work as few as 16 hours a week. This suggests the salaries may be recorded as **full-time equivalents** rather than actual pay, so it would be worth checking with HR before drawing conclusions. Good EDA questions what the data really means.

Choose **your best chart** from this task, or create a new one, and make it **report-ready**:

## 🔴 Extension 2 — Reducing skew with a log transform

`np.log()` squeezes large values closer together, which can make a right-skewed variable more symmetric. Create a column `log_salary` with `np.log()`, then compare the skew of `salary` and `log_salary` and draw a histogram of `log_salary`.

````python
# 🔴 Extension 2: a log transform of salary
df['log_salary'] = np.log(df['salary'])

print(f"Skew of salary:     {df['salary'].skew():.2f}")
print(f"Skew of log_salary: {df['log_salary'].skew():.2f}")

plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='log_salary', bins=20, kde=True)
plt.title('Distribution of log(salary)')
plt.xlabel('log(salary)')
plt.ylabel('Number of staff')
plt.show()

# ✅ Skew falls from 0.88 to 0.46: closer to symmetric
````

## 🔴 Extension 3 — Fix the misleading chart

The chart below compares **average** salary by department, but it is **misleading**. Run it, find the problem, and create a fixed version in the next cell. Explain the problem in a comment.

````python
# 🔴 Extension 3: a misleading chart — run it and look closely
dept_means = df.groupby('department')['salary'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 5))
plt.bar(dept_means.index, dept_means.values, color='steelblue')
plt.ylim(23000, 39000)
plt.title('Average salary by department')
plt.ylabel('Salary (£)')
plt.show()
````

````python
# 🔴 Extension 3: your fixed chart
dept_means = df.groupby('department')['salary'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 5))
plt.bar(dept_means.index, dept_means.values, color='steelblue')
plt.title('Average salary by department')
plt.ylabel('Salary (£)')
plt.show()

# What was misleading?
# The y-axis started at £23,000 instead of £0, so the bars exaggerated the differences:
# Finance's bar looked about ten times taller than Customer Service's, when its average
# salary is only about 1.5 times higher (£38,038 against £24,579). Bar charts should start at zero.
````
