## Section 1 (Load): Load the data

Fill in the blanks to import pandas and load the data. The web address to the `retailco_customer_transactions.csv` file saved in Github is already typed for you.

```python
# ➡️ Task 1.1: import pandas and load the data
import pandas as pd

url = "https://raw.githubusercontent.com/7000lbsbai/7000LBSBAI/main/datasets/retailco_customer_transactions.csv"
df = pd.read_csv(url)

print(f"Loaded {df.shape[0]} rows and {df.shape[1]} columns")

# ✅ Expected: Loaded 200 rows and 9 columns
```

---
## Section 2 (Inspect): First look

> 📝 **Tip:** each of these goes in its own cell, because a cell only displays its **last** result.

```python
# ➡️ Task 2.1: show the number of rows and columns
df.shape

# ✅ Expected: (200, 9)
```

```python
# ➡️ Task 2.2: show the first 10 rows
df.head(10)

# ✅ Expected: rows 0 to 9, starting with CUST-0118 (Notebook Set, £17.98)
```

```python
# ➡️ Task 2.3: show the last 5 rows
df.tail()

# ✅ Expected: rows 195 to 199
```

### 🔴 Extension 2.4

Show 5 **random** rows using `df.sample(5)`. Run the cell twice — what changes?

```python
# 🔴 Extension 2.4: 5 random rows
df.sample(5)

# ✅ Different rows each time, because they are chosen at random
```

---
## Section 3 (Inspect): Structure and missing values

```python
# ➡️ Task 3.1: show the column names, data types and non-null counts
df.info()

# ✅ Expected: 9 columns, 200 non-null in each.
#    unit_price and total_spend are float64, quantity is int64,
#    the other 6 columns are text (object, or str in newer versions of pandas)
```

```python
# ➡️ Task 3.2: count the missing values in each column
df.isna().sum()

# ✅ Expected: 0 in every column
```

```python
# ➡️ Task 3.3: list the column names
df.columns

# ✅ Expected: customer_id, date, product_category, product_name, unit_price,
#    quantity, total_spend, region, payment_method
```

---
## Section 4 (Inspect): Summary statistics

```python
# ➡️ Task 4.1: summary statistics for the number columns
df.describe()

# ✅ Expected: summaries of unit_price, quantity and total_spend
```

### ➡️ Task 4.2 — Statistics for one column

Select the `total_spend` column, then use `.mean()`, `.median()`, `.max()` and `.min()`. The f-strings from Week 1 then print each value as money.

```python
# ➡️ Task 4.2: statistics for total_spend
mean_spend   = df['total_spend'].mean()
median_spend = df['total_spend'].median()
max_spend    = df['total_spend'].max()
min_spend    = df['total_spend'].min()

print(f"Mean spend:    £{mean_spend:.2f}")
print(f"Median spend:  £{median_spend:.2f}")
print(f"Largest spend: £{max_spend:.2f}")
print(f"Smallest spend: £{min_spend:.2f}")

# ✅ Expected: Mean £81.42, Median £39.98, Largest £949.95, Smallest £3.49
```

### ➡️ Task 4.3 — Total revenue

Add up the `total_spend` column and print it as £, with a comma for the thousands.

> 📝 **Tip:** if you display a total without formatting, you may see something like `16283.470000000001`. Tiny rounding differences like this are normal in computers. Formatting with `:,.2f` shows it properly as `16,283.47`.

```python
# ➡️ Task 4.3: total revenue
total_revenue = df['total_spend'].sum()

print(f"Total revenue: £{total_revenue:,.2f}")

# ✅ Expected: Total revenue: £16,283.47
```

### 🔴 Extension 4.4

`df.describe()` only covers the number columns. Try `df.describe(include='object')` to summarise the text columns. What does the `'top'` row show?

```python
# 🔴 Extension 4.4: summarise the text columns
df.describe(include='object')

# ✅ 'top' is the most common value in each column,
#    e.g. Furniture (63) for product_category and Credit Card (72) for payment_method
```

## Section 5 (Select): Selecting columns

- **One column:** `df['column_name']`
- **Several columns:** `df[['column_1', 'column_2']]` — note the double brackets
- **Count each value in a column:** `df['column_name'].value_counts()`

```python
# ➡️ Task 5.1: select the region column
df['region']
```

```python
# ➡️ Task 5.2: select the customer_id, product_name and total_spend columns
df[['customer_id', 'product_name', 'total_spend']]
```

```python
# ➡️ Task 5.3: count the transactions in each product category
df['product_category'].value_counts()

# ✅ Expected: Furniture 63, Stationery 55, Electronics 46, Software 36
```

```python
# ➡️ Task 5.4: count the transactions for each payment method
df['payment_method'].value_counts()

# ✅ Expected: Credit Card 72, Debit Card 60, PayPal 41, Bank Transfer 17, Voucher 10
```

### 🔴 Extension 5.5

How many **different** customers are there? Use `.nunique()` on the `customer_id` column.

```python
# 🔴 Extension 5.5: number of different customers
df['customer_id'].nunique()

# ✅ Expected: 200 — every customer appears once
```

### 🔴 Extension 5.6

`df.iloc[12]` shows the row at position 12. `df.loc[12, 'product_name']` shows one value from it. What did the customer in row 12 buy, and how much did they spend?

```python
# 🔴 Extension 5.6: look up row 12
print(df.iloc[12])
print()
print("Customer bought: " + df.loc[12, 'product_name'])
print(f"Customer spent: £{df.loc[12, 'total_spend']:,.2f}")

# ✅ Expected: Desk Lamp, 69.0 (CUST-0060)
# Note the use of different print statements: one concatenates text and value,
# the other uses f-string format.
```
