## The scenario

The manager of **BeanHouse Coffee** has sent you three questions about last week's sales. You will use the same 60 orders as the code-along.

## Getting started — load the data

This is a new notebook, so the data needs loading again. The code is written for you: just run it.

```python
# ✍️ Load the data
import pandas as pd

url = "https://raw.githubusercontent.com/7000lbsbai/7000LBSBAI/main/datasets/beanhouse_coffee_sales.csv"
df = pd.read_csv(url)

# Create the total column, as in the code-along
df['total'] = df['price'] * df['quantity']

print(f"Loaded {df.shape[0]} orders")
```

---
## ➡️ Question 1 — How many orders were paid in cash?

**Steps:**
1. Filter to keep only the rows where `payment_method` is `'Cash'`
2. Count the rows with `len()`
3. Print the answer

```python
# ➡️ Question 1
# Step 1: filter to the cash orders
cash_orders = df[df['payment_method'] == 'Cash']

# Step 2: count them
number_of_cash_orders = len(cash_orders)

# Step 3: print the answer
print(f"Orders paid in cash: {number_of_cash_orders}")

# ✅ Expected: Orders paid in cash: 14
```

---
## ➡️ Question 2 — How much revenue did each branch make?

**Steps:**
1. Group by `branch`, take the `total` column and add it up
2. Sort from highest to lowest
3. Display the result

> 📝 **Hint:** read it as a sentence — *"for each branch, take the total and add it up"*.

```python
# ➡️ Question 2
branch_revenue = df.groupby('branch')['total'].sum()

branch_revenue = branch_revenue.sort_values(ascending=False)

branch_revenue

# ✅ Expected: City Centre 169.8, University 160.8, Station 60.8
```

---
## ➡️ Question 3 — Which 5 orders had the highest total?

Show only the `order_id`, `branch`, `item`, `quantity` and `total` columns.

> 📝 **Hint:** `nlargest` needs a number and a column name.

```python
# ➡️ Question 3
top_orders = df.nlargest(5, 'total')

top_orders[['order_id', 'branch', 'item', 'quantity', 'total']]

# ✅ Expected (top 3): B041 City Centre Iced Latte 6 £22.20,
#                     B034 University Flat White 6 £21.00,
#                     B011 City Centre Muffin 6 £17.40
```

---
## 🔴 Extension 1 — How much revenue did Food make at the City Centre branch?

This time you write the code. Use **two filters, one after the other**:
1. Filter `df` to the `'City Centre'` branch and save it as `city_centre`
2. Filter `city_centre` to the `'Food'` category and save it as `city_centre_food`
3. Add up the `total` column of `city_centre_food`
4. Print the answer, formatted as money: `£{value:.2f}`

```python
# 🔴 Extension 1
# Step 1: City Centre orders
city_centre = df[df['branch'] == 'City Centre']

# Step 2: City Centre Food orders
city_centre_food = city_centre[city_centre['category'] == 'Food']

# Step 3: add up the total
food_revenue = city_centre_food['total'].sum()

# Step 4: print the answer
print(f"City Centre Food revenue: £{food_revenue:.2f}")

# ✅ Expected: City Centre Food revenue: £31.30 (from 4 orders)
# 📝 Please note: in Step 2 you must filter city_centre, not df —
#    filtering df again loses the City Centre filter (Food revenue for all branches is £61.50).
```

---
## 🔴 Extension 2 — What was the best-selling item?

Which item sold the most units (the highest total `quantity`) across the week?

> 📝 **Hint:** group by `item`, add up `quantity`, then sort from highest to lowest.

```python
# 🔴 Extension 2: best-selling item by quantity
item_sales = df.groupby('item')['quantity'].sum().sort_values(ascending=False)
item_sales.head()

# ✅ Expected: Flat White 24, Americano 14, Tea 13, Iced Latte 12 ...
```

---
## 🔴 Extension 3 — Which day of the week was busiest?

Which date had the highest total revenue?

```python
# 🔴 Extension 3: revenue by date
daily_revenue = df.groupby('date')['total'].sum().sort_values(ascending=False)
daily_revenue

# ✅ Expected: 2024-03-08 (a Friday) with £103.10 — about double any other day
```
