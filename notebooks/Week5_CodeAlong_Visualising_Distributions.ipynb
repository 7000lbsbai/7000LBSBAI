## Recognising distribution shapes

| Shape | What it looks like | What it means for your EDA |
|-------|-------------------|----------------------------|
| **Symmetric** | Roughly the same on both sides; mean ≈ median | The mean is a fair summary |
| **Right-skewed** | A long tail to the right; mean > median | Report the median; look at the large values |
| **Left-skewed** | A long tail to the left; mean < median | Report the median; look at the small values |
| **Two peaks** | Two separate humps | Two groups are probably mixed together —> split them and analyse separately |

**Skew as a number** (`.skew()`): roughly between −0.5 and 0.5 is close to symmetric; above 1 (or below −1) is strongly skewed.

## Set up — run this cell

This week we add a fourth package: **seaborn** (`sns`), which makes statistical charts in one line. It is built on matplotlib, so `plt.title()`, `plt.xlabel()`, `plt.ylabel()` and `plt.show()` still work.

````python
# ✍️ Set up —> just run this cell
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')     # a clean look for every chart

url = "https://raw.githubusercontent.com/7000lbsbai/7000LBSBAI/main/datasets/beanhouse_daily_sales.csv"
df = pd.read_csv(url)
print(f"Data loaded: {df.shape[0]} rows")
````

---
# Block 1: The shape of a distribution
*about 15 minutes*

A **histogram** shows how often each range of values occurs. Adding `kde=True` draws a smooth **density curve** over the bars, which makes the shape easier to see.

In this cell we also mark the **mean** and **median**, and print the **skew**.

````python
# 👀 Block 1: a histogram of daily revenue, with the mean, median and skew
# Watch the demo first, then run this cell yourself

plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='revenue', bins=25, kde=True)
plt.axvline(df['revenue'].mean(), color='red', linestyle='--', label=f"Mean £{df['revenue'].mean():.0f}")
plt.axvline(df['revenue'].median(), color='black', label=f"Median £{df['revenue'].median():.0f}")

plt.title('Daily revenue — all branches')
plt.xlabel('Revenue (£)')
plt.ylabel('Number of days')
plt.legend()
plt.show()

print(f"Skew: {df['revenue'].skew():.2f}")
````

> 🔎 **Try it:** change `bins=25` to `bins=8`, then to `bins=60`. How does the shape change? Too few bins hide detail; too many make it noisy.

# Block 2: Comparing groups
*about 15 minutes*

To compare distributions across groups, two charts work well together:

- A **boxplot** shows each group's median, middle 50% (the box) and outliers.
- Overlapping **histograms** with `hue='branch'` show each group's full shape in a different colour.

`plt.subplots(1, 2)` creates **two charts side by side**. Each seaborn chart is placed with `ax=axes[0]` or `ax=axes[1]`, and each one gets its own title with `axes[0].set_title()`.

````python
# 👀 Block 2: a boxplot and overlapping histograms, side by side
# Watch the demo first, then run this cell yourself

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Left: a boxplot for each branch
sns.boxplot(data=df, x='branch', y='revenue', ax=axes[0])
axes[0].set_title('Daily revenue by branch')
axes[0].set_xlabel('')
axes[0].set_ylabel('Revenue (£)')

# Right: overlapping histograms, one colour per branch
sns.histplot(data=df, x='revenue', hue='branch', bins=30, element='step', ax=axes[1])
axes[1].set_title('The shape of each branch')
axes[1].set_xlabel('Revenue (£)')
axes[1].set_ylabel('Number of days')

plt.tight_layout()     # stops the two charts overlapping
plt.show()

print(df.groupby('branch')['revenue'].skew().round(2))
````

# Block 3: Hidden groups
*about 15 minutes*

One histogram **per branch** makes each shape easier to see than overlapping colours.

This cell uses a **for loop** (Week 1!) to draw the three charts. `zip(axes, branches)` pairs each chart position with a branch name, so the loop draws one histogram in each position. `sharex=True` gives all three charts the same x-axis, so they are easy to compare.

````python
# 👀 Block 3: one histogram per branch
# Watch the demo first, then run this cell yourself

branches = ['City Centre', 'University', 'Station']
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5), sharex=True)

for ax, branch in zip(axes, branches):
    branch_data = df[df['branch'] == branch]
    sns.histplot(data=branch_data, x='revenue', bins=15, kde=True, ax=ax)
    ax.set_title(branch)
    ax.set_xlabel('Revenue (£)')
    ax.set_ylabel('Number of days')

plt.tight_layout()
plt.show()
````

# Block 4: Categories, and a chart ready for a report
*about 15 minutes*

A **count plot** shows how many rows fall into each category —> the distribution of a **categorical** variable.

This cell also shows how to make a chart **report-ready** and save it as an image with `plt.savefig()`.

> ⚠️ Always call `plt.savefig()` **before** `plt.show()`. After `show()` the chart is cleared, and you save a blank image.

````python
# 👀 Block 4: a report-ready count plot of busy days, saved as an image
# Watch the demo first, then run this cell yourself

# Flag the busy days: the top 25% of days (np.where from Week 4)
df['busy'] = np.where(df['revenue'] > df['revenue'].quantile(0.75), 'Busy', 'Normal')
busy_days = df[df['busy'] == 'Busy']

day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

plt.figure(figsize=(10, 5))
sns.countplot(data=busy_days, x='day', order=day_order, color='steelblue')
plt.title('When are BeanHouse\'s busiest days? (top 25% of days by revenue)')
plt.xlabel('')
plt.ylabel('Number of busy days')
plt.tight_layout()

plt.savefig('busy_days.png', dpi=150)     # save BEFORE show
plt.show()
````
