## Section 1 (Load): Load the data

This is a new notebook, so the data needs loading again. Write the two missing lines yourself: import pandas, then read the CSV from `url`.

> 📝 **Hint:** look back at Task 1 Notebook - Task 1.1 if you need to.

````python
# ➡️ Task 1.1: import pandas and load the data
url = "https://raw.githubusercontent.com/7000lbsbai/7000LBSBAI/main/datasets/retailco_customer_transactions.csv"

# Write your two lines below
import pandas as pd
df = pd.read_csv(url)

print(f"Loaded {df.shape[0]} rows and {df.shape[1]} columns")

# ✅ Expected: Loaded 200 rows and 9 columns
````

---
## Section 2: Sort

- `df.sort_values('column', ascending=False)` sorts largest first
- `df.nlargest(5, 'column')` and `df.nsmallest(5, 'column')` give the top or bottom 5 rows

````python
# ➡️ Task 2.1: sort by total_spend, largest first, and show the top 10
sorted_df = df.sort_values('total_spend', ascending=False)

sorted_df[['customer_id', 'product_name', 'total_spend']].head(10)

# ✅ Expected: first row CUST-0098, Ergonomic Chair, 949.95
````

````python
# ➡️ Task 2.2: the 5 transactions with the highest total_spend, using nlargest
df.nlargest(5, 'total_spend')

# ✅ Expected: 949.95, 569.97, 569.97, 499.90, 449.90
````

````python
# ➡️ Task 2.3: the 5 transactions with the lowest unit_price
df.nsmallest(5, 'unit_price')

# ✅ Expected: Sticky Notes at £3.49
````

### 🔴 Extension 2.4 — Sort by two columns

Pass a list of columns, and a list of `True`/`False` values for `ascending`. This sorts by category A–Z, then by spend from largest to smallest within each category:

```python
df.sort_values(['product_category', 'total_spend'], ascending=[True, False])
```

Run it, then change it to sort by `region` instead of `product_category`.

---
## Section 3: Filter

- `df[df['column'] > 100]` keeps the rows where the condition is True
- Use `==` to match text, in quotes: `df[df['region'] == 'London']`
- `len()` counts the rows in the result

````python
# ➡️ Task 3.1: keep only the transactions over £100
premium = df[df['total_spend'] > 100]

print(f"Transactions over £100: {len(premium)}")

# ✅ Expected: Transactions over £100: 42
````

````python
# ➡️ Task 3.2: the total value of the transactions over £100
premium_value = premium['total_spend'].sum()

print(f"Value of transactions over £100: £{premium_value:,.2f}")

# ✅ Expected: Value of transactions over £100: £10,336.67
````

````python
# ➡️ Task 3.3: how many transactions came from London?
london = df[df['region'] == 'London']

print(f"London transactions: {len(london)}")

# ✅ Expected: London transactions: 25
````

````python
# ➡️ Task 3.4: how many Furniture transactions were there?
furniture = df[df['product_category'] == 'Furniture']

print(f"Furniture transactions: {len(furniture)}")

# ✅ Expected: Furniture transactions: 63
````

````python
# ➡️ Task 3.5: how many transactions had a quantity of 5 or more? Print the answer.
bulk = df[df['quantity'] >= 5]

print(f"Transactions of 5 or more units: {len(bulk)}")

# ✅ Expected: Transactions of 5 or more units: 17
````

### 🔴 Extension 3.6 — Two conditions at once

To filter on two conditions, use `&` (and), and put **each** condition in its own brackets:

```python
df[(condition_1) & (condition_2)]
```

How many **Furniture** transactions were **over £100**?

````python
# 🔴 Extension 3.6: Furniture transactions over £100
big_furniture = df[(df['product_category'] == 'Furniture') & (df['total_spend'] > 100)]

print(f"Furniture transactions over £100: {len(big_furniture)}")

# ✅ Expected: 23
# 📝 Please note: common errors are using 'and' instead of & (ValueError)
#    and leaving out the brackets around each condition
````

## Section 4: Group

`df.groupby('group_column')['value_column'].sum()` → *"for each group, add up the values"*

Swap `.sum()` for `.mean()` to get the average.

````python
# ➡️ Task 4.1: total revenue for each product category, highest first
category_revenue = df.groupby('product_category')['total_spend'].sum()

category_revenue.sort_values(ascending=False)

# ✅ Expected: Furniture 8727.99, Software 3859.07, Electronics 2544.18, Stationery 1152.23
````

````python
# ➡️ Task 4.2: total revenue for each region, highest first
df.groupby('region')['total_spend'].sum().sort_values(ascending=False)

# ✅ Expected: Midlands 3129.83 first, South West 1127.72 last
````

````python
# ➡️ Task 4.3: AVERAGE total_spend for each payment method, highest first
df.groupby('payment_method')['total_spend'].mean().sort_values(ascending=False)

# ✅ Expected: Voucher 99.68, Credit Card 95.19, Bank Transfer 89.86, PayPal 71.22, Debit Card 66.43
````

### 🔴 Extension 4.4 — Several statistics at once

`.agg()` calculates several statistics in one go. Run this, then add `'max'` to the list:

```python
df.groupby('product_category')['total_spend'].agg(['count', 'sum', 'mean'])
```

````python
# 🔴 Extension 4.4: several statistics for each category
df.groupby('product_category')['total_spend'].agg(['count', 'sum', 'mean', 'max'])
````

## Section 5: Create new columns

`df['new_column'] = calculation` works out the calculation for **every row at once**.

This is the same arithmetic you used in Week 1 Task 2 but now applied to whole columns.

````python
# ➡️ Task 5.1: create a revenue column = unit_price × quantity
df['revenue'] = df['unit_price'] * df['quantity']

df[['product_name', 'unit_price', 'quantity', 'revenue']].head()

# ✅ Expected first row: Notebook Set, 8.99, 2, 17.98
````

````python
# ➡️ Task 5.2: create a vat column = revenue × 0.20
df['vat'] = df['revenue'] * 0.20

df[['product_name', 'revenue', 'vat']].head()

# ✅ Expected first row: revenue 17.98, vat 3.596
````

````python
# ➡️ Task 5.3: create a revenue_inc_vat column = revenue + vat, and show the first 5 rows
df['revenue_inc_vat'] = df['revenue'] + df['vat']

df[['product_name', 'revenue', 'vat', 'revenue_inc_vat']].head()

# ✅ Expected first row: 17.98, 3.596, 21.576
````

### 🔴 Extension 5.4 — Check your calculation

Your `revenue` column should match the `total_spend` column exactly. This code compares them row by row. Run it.

Do all 200 rows match?

````python
# 🔴 Extension 5.4: does revenue match total_spend?
matches = df['revenue'].round(2) == df['total_spend'].round(2)

print(f"Rows that match: {matches.sum()} out of {len(df)}")

# ✅ Expected: Rows that match: 200 out of 200
# 📝 Please note: .round(2) avoids tiny floating-point differences making equal values look unequal
````

## 🔴 Extension 6 — Check the AI's code

A colleague asked an AI tool: *"How many transactions came from London customers who paid by Credit Card?"*

The AI wrote the code below. It runs without an error - but is it the right answer?

1. Run the cell.
2. You found in Task 3.3 that there are only **25** London transactions in total. Does the AI's answer make sense?
3. Find the mistake, fix it in the next cell, and write one sentence explaining what went wrong.

````python
# 🔴 AI-generated code — run it and check the answer
london_card = df[(df['region'] == 'London') | (df['payment_method'] == 'Credit Card')]

print(f"London Credit Card transactions: {len(london_card)}")
````
