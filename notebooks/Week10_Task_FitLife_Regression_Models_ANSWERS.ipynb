## Set up: run this cell

This cell loads and cleans FitLife with the Week 8 function, and creates `cancelled_flag` from `cancel_date`, as you did in Week 9.

````python
# ✍️ Set up: just run this cell
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf

# Set the Seaborn chart style
sns.set_theme(style='whitegrid')

# Create a function to clean the raw FitLife member data
def clean_members(data):
    """Clean the FitLife member data (the Week 8 function)."""
    data = data.copy()
    data = data.drop_duplicates()
    data['membership_type'] = data['membership_type'].str.strip().str.title()
    data['monthly_fee'] = pd.to_numeric(data['monthly_fee'].str.replace('£', ''))
    data = data[(data['age'] >= 18) & (data['age'] <= 100)]
    data['visits_per_month'] = data['visits_per_month'].fillna(data['visits_per_month'].median())
    return data

# Import the FitLife Members data
raw = pd.read_csv("https://raw.githubusercontent.com/7000lbsbai/7000LBSBAI/main/datasets/fitlife_members.csv")

# Run the clean_members function on the raw dataset
df = clean_members(raw)

# The outcome, created from the date (Week 9): 1 if the member has a cancel date
df['cancelled_flag'] = np.where(df['cancel_date'].notna(), 1, 0)

# Check all processed correctly
print(f"Data ready: {df.shape[0]} members, {df['cancelled_flag'].mean() * 100:.1f}% cancelled")
````

---
# Section A: What drives spend?
*about 15 minutes*

In Week 9, visits alone explained only about **24%** of the variation in spend (R² = 0.24). Now include several predictors at once.

> 📝 **Hint:** code-along Block 1. Membership type is a category, so use `C(membership_type)`. The reference group is **Annual** (first alphabetically).

````python
# ➡️ A.1 Fit a linear model of monthly_spend explained by visits_per_month, classes_per_month, pt_sessions_per_month and C(membership_type).
# Show a table of the coefficients, p-values and 95% confidence intervals, and print the adjusted R²
spend_model = smf.ols('monthly_spend ~ visits_per_month + classes_per_month + pt_sessions_per_month + C(membership_type)', data=df).fit()

spend_results = pd.DataFrame({
    'coefficient': spend_model.params,
    'p_value': spend_model.pvalues,
    'lower_95': spend_model.conf_int()[0],
    'upper_95': spend_model.conf_int()[1]
}).round(3)
print(spend_results)
print(f"Adjusted R²: {spend_model.rsquared_adj:.3f}")

# ✅ Visits £1.85, classes £3.04, personal training £20.04 per session; Monthly Flex +£7.50 and
#    Off-Peak -£4.53 compared with Annual. All p < 0.05. Adjusted R² 0.881
````

### 💬 What does it mean?

- Explain the coefficients for personal training and for Monthly Flex in plain English.
- Why is the coefficient for visits smaller here (£1.85) than in Week 9 (£2.67)?
- How much better is this model than the Week 9 model?

**✏️ MODEL ANSWER:**

- Holding the other factors constant:
  - Each **personal training session** a month adds about **£20** to monthly spend
  - Each **class** adds about **£3**
  - Each **visit** adds about **£1.85**.
  - **Monthly Flex** members spend about **£7.50** more than Annual members
  - **Off-Peak** members spend about **£4.50** less, which reflects their different fees.
- All are significant (p < 0.05).
- The visits coefficient is **smaller** than in Week 9 because members who visit more also tend to take more classes: on its own, visits was picking up some of the effect of classes. Holding classes constant separates the two.
- The adjusted R² rose from **0.24 to 0.88**: the model now explains almost **90%** of the variation in spend, and personal training has the largest estimated effect on monthly spending.

# Section B: Choosing predictors
*about 10 minutes*

Each membership type has a standard fee, and some members pay a discounted fee. So `monthly_fee` is **largely decided by** `membership_type`. What happens if you include both?

> 📝 **Hint:** code-along Block 3.

````python
# ➡️ B.1 Fit the A.1 model again with monthly_fee added.
# Print the coefficients and p-values. What happens to the membership type coefficients?
fee_model = smf.ols('monthly_spend ~ visits_per_month + classes_per_month + pt_sessions_per_month + C(membership_type) + monthly_fee', data=df).fit()

print(pd.DataFrame({'coefficient': fee_model.params, 'p_value': fee_model.pvalues}).round(3))

# ✅ Monthly Flex falls from +£7.50 to +£0.58 (p = 0.061), Off-Peak from -£4.53 to +£0.07 (p = 0.82):
#    once the fee is in the model, membership type seems to have almost no effect
````

### ✏️ B.2 What happened, and what would you do?

- Why did the membership type coefficients shrink and stop being significant when the fee was added?
- Which version would you report, and why?
- A colleague suggests adding `cancel_date` to a model of **cancellation**. Why would that be a mistake?

**MODEL ANSWER:**

- Fee and membership type **multicollinearity**: the fee is mostly decided by the membership type, so they carry much of the same information. Once both are in the model, the effect is credited to the fee, and membership type appears to have **almost no effect**.
- Both versions fit well, but they answer different questions. I would report **one** of them and explain why: membership type is easier for a manager to act on, or the fee if pricing is the focus. Including both would be confusing and could be misleading.

- Adding `cancel_date` to a model of cancellation would be **leakage**: our outcome was *created from* `cancel_date`, so the model would simply rediscover our own definition. It would look perfect and tell us nothing.

# Section C: What makes members cancel?
*about 20 minutes*

Now create a new model **cancellation**, a yes/no outcome. Use these predictors: `visits_per_month`, `classes_per_month`, `distance_km`, `age`, `C(membership_type)` and `C(home_gym)`.

> 📝 **Hint:** code-along Block 2. Use `smf.logit(...)` and `.fit(disp=0)`, then `np.exp()` for odds ratios.

````python
# ➡️ C.1 Fit the logistic model.
# Show a table of odds ratios with 95% confidence intervals and p-values,
# without the intercept, rounded to 3 decimal places
cancel_model = smf.logit('cancelled_flag ~ visits_per_month + classes_per_month + distance_km + age + C(membership_type) + C(home_gym)', data=df).fit(disp=0)

odds = pd.DataFrame({
    'odds_ratio': np.exp(cancel_model.params),
    'lower_95': np.exp(cancel_model.conf_int()[0]),
    'upper_95': np.exp(cancel_model.conf_int()[1]),
    'p_value': cancel_model.pvalues
}).round(3).drop(index='Intercept')
print(odds)

# ✅ Monthly Flex 5.54 and Off-Peak 1.95 (against Annual); visits 0.824; classes 0.885; distance 1.075;
#    Northgate 1.36 (p = 0.01); Riverside 0.95 (p = 0.62); age 0.991 (p = 0.052)
````

```python
members = pd.DataFrame({

```

````python
# ➡️ C.2 The predicted chance of cancelling for each example member
members = pd.DataFrame({
    'description': ['Monthly Flex, rarely visits, lives far away', 'Monthly Flex, visits often, lives nearby',
                    'Annual, rarely visits, lives far away', 'Annual, visits often, lives nearby'],
    'visits_per_month': [4, 12, 4, 12],
    'classes_per_month': [1, 3, 1, 3],
    'distance_km': [8, 2, 8, 2],
    'age': [30, 30, 30, 30],
    'membership_type': ['Monthly Flex', 'Monthly Flex', 'Annual', 'Annual'],
    'home_gym': ['City Centre', 'City Centre', 'City Centre', 'City Centre']
})
members['chance_of_cancelling_%'] = (cancel_model.predict(members) * 100).round(1)
print(members[['description', 'chance_of_cancelling_%']])

# ✅ 63.9%, 16.0%, 24.2% and 3.3%
````

### 💬 What does it mean?

- Which factors increase the chance of cancelling, and which reduce it? Use the odds ratios.
- Which factors are **not** significant once everything is taken into account?
- Use the predicted chances to explain the results to the manager in one or two sentences.

**✏️ MODEL ANSWER:**

- Holding the other factors constant:
  - **Monthly Flex** members have about **5.5 times** the odds of cancelling of Annual members
  - **Off-Peak** members have about **twice** the odds
  - Each extra **visit** a month cuts the odds by about **18%** (odds ratio 0.82)
  - Each extra **class** cuts the odds by about **12%**
  - Each extra **kilometre** of distance **raises** the odds by about **7.5%**.
  - Members of **Northgate** have about **36%** higher odds than City Centre (p = 0.01)
- **Riverside** is no different from City Centre (p = 0.62)
- **Age** is just above the 0.05 cut-off (p = 0.052), so on this model it does not look significant (but see Section D).
- For the manager: *a Monthly Flex member who rarely visits and lives 8 km away has about a 64% chance of cancelling, but an Annual member who visits often and lives nearby has only about 3%*.

# Section D: Is a straight line right for age?
*about 10 minutes*

In C.1, age was **not quite significant** (p = 0.052). But in Week 8 you found a **U-shape**: the **youngest** and **oldest** members cancelled most. A single `age` term assumes the effect is a **straight line** (the chance steadily rises or falls with age), so it can miss a U-shape.

Replace `age` with **age bands**, and choose the **35-49** band as the reference group (it had the lowest cancellation rate in Week 8):

```python
df['age_band'] = pd.cut(df['age'], bins=[17, 24, 34, 49, 100], labels=['18-24', '25-34', '35-49', '50+'])
```

To choose a reference group yourself, use `Treatment()` inside `C()`:

```python
C(age_band, Treatment('35-49'))
```

````python
# ➡️ D.1 Create age_band, then fit the C.1 model with C(age_band, Treatment('35-49')) instead of age.
# Print the odds ratios and p-values for the age bands only
df['age_band'] = pd.cut(df['age'], bins=[17, 24, 34, 49, 100], labels=['18-24', '25-34', '35-49', '50+'])

band_model = smf.logit("cancelled_flag ~ visits_per_month + classes_per_month + distance_km + C(age_band, Treatment('35-49')) + C(membership_type) + C(home_gym)", data=df).fit(disp=0)

band_odds = pd.DataFrame({'odds_ratio': np.exp(band_model.params), 'p_value': band_model.pvalues}).round(3)
print(band_odds[band_odds.index.str.contains('age_band')])

# ✅ 18-24: 1.798 (p < 0.001); 25-34: 1.077 (p = 0.53); 50+: 1.399 (p = 0.034)
````

> 📝 The formula uses **double** quotes on the outside, because `'35-49'` already uses single quotes.

### 💬 What does it mean?

- What does the model with age bands show that the straight-line model missed?
- What does this tell you about choosing the right form for a predictor in your own models?

**✏️ MODEL ANSWER:**

- With bands, age clearly **does** matter: compared with members aged 35-49, the **18-24** group has about **80% higher** odds of cancelling (p < 0.001) and the **50+** group about **40% higher** odds (p = 0.03), while 25-34 is no different.
- That is the **U-shape** from Week 8. The straight-line `age` term missed it because the effect goes **up at both ends**: a single slope cannot show that, so it came out close to zero.
- Before trusting a model, **check the shape** of each relationship with the techniques from earlier weeks (rates by band, charts). If it is not a straight line, use bands (or another form) in the model.

````python
# 🤖 E.1 Your reviewed version (it must give the same numbers)

def odds_ratio_table(model):
    """Odds ratios, 95% confidence intervals and p-values for a logistic model, without the intercept."""
    table = pd.DataFrame({
        'odds_ratio': np.exp(model.params),
        'lower_95': np.exp(model.conf_int()[0]),
        'upper_95': np.exp(model.conf_int()[1]),
        'p_value': model.pvalues
    })
    return table.drop(index='Intercept').round(3)

# The cancellation model from C.1, then its table of odds ratios
cancel_odds = odds_ratio_table(cancel_model)
print(cancel_odds)

# Check: the same numbers as C.1
print(f"Unchanged: {cancel_odds.equals(odds)}")
````

### ✏️ F.2 Which model for which question?

- In your own words, how do you decide whether to use linear or logistic regression?
- What would you check **before** choosing the predictors for a model?

**MODEL ANSWER:**

I look at the **outcome**. If it is a **number** (such as spend or delivery time), I use **linear** regression and interpret the coefficients as the change in the outcome for each extra unit of a predictor, holding the others constant. If it is **yes or no** (such as cancelled or churned), I convert it to 1/0 and use **logistic** regression, interpreting **odds ratios** and **predicted probabilities**.

Before choosing predictors, I check that none is **made from the outcome** or only known after it (**leakage**), check for **multicollinearity** among the predictors (a correlation heatmap, and thinking about how each column was created), and that each relationship has the **shape** the model assumes, using bands where it does not (as with age).

## 🔴 Extension 2: building a model step by step

Use a **loop** to fit linear models of `monthly_spend`, adding one predictor at a time, and print the adjusted R² each time. Which predictor adds the most?

Start with: `['pt_sessions_per_month', 'classes_per_month', 'visits_per_month', 'C(membership_type)', 'distance_km']`

````python
# 🔴 Extension 2: adding predictors one at a time
predictors = ['pt_sessions_per_month', 'classes_per_month', 'visits_per_month', 'C(membership_type)', 'distance_km']

for i in range(1, len(predictors) + 1):
    formula = 'monthly_spend ~ ' + ' + '.join(predictors[:i])
    m = smf.ols(formula, data=df).fit()
    print(f"{m.rsquared_adj:.3f}   {formula}")

# Personal training alone gives about 0.50; classes and visits add a lot; membership type adds more;
# distance adds almost nothing
````
