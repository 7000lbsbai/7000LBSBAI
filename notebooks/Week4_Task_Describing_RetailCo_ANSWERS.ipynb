## Section 1: Load and check the data

```python
# ➡️ Task 1.1: import pandas, NumPy and matplotlib, then load the data
url = "https://raw.githubusercontent.com/7000lbsbai/7000LBSBAI/main/datasets/retailco_customer_transactions.csv"

# Write your code below this line
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(url)
print("Data loaded!")
```

> 📝 **Hint:** Code-along Step 1a.

```python
# ➡️ Task 1.2: print the shape, the total number of missing values, and the number of duplicate rows
print(f"Rows and columns: {df.shape}")
print(f"Missing values:   {df.isnull().sum().sum()}")
print(f"Duplicate rows:   {df.duplicated().sum()}")

# ✅ (200, 9), 0 and 0 — the data is clean
```

> 📝 **Hint:** Code-along Step 1b. These are some of the Week 3 checks.

## Section 2: Describe transaction values

The `total_spend` column holds the value of each transaction.

```python
# ➡️ Task 2.1: print the mean and median of total_spend, formatted as £0.00
mean_spend = df['total_spend'].mean()
median_spend = df['total_spend'].median()

print(f"Mean transaction:   £{mean_spend:.2f}")
print(f"Median transaction: £{median_spend:.2f}")

# ✅ Mean £81.42, median £39.98
```

> 📝 **Hint:** Code-along Step 2a.

```python
# ➡️ Task 2.2: print the standard deviation, the smallest and largest transaction, and the range
low = df['total_spend'].min()
high = df['total_spend'].max()

print(f"Standard deviation: £{df['total_spend'].std():.2f}")
print(f"Smallest: £{low:.2f}")
print(f"Largest:  £{high:.2f}")
print(f"Range:    £{high - low:.2f}")

# ✅ Std £118.43, smallest £3.49, largest £949.95, range £946.46
```

> 📝 **Hint:** Code-along Steps 2b and 2c.

```python
# ➡️ Task 2.3: show the quartiles of total_spend, then print the IQR
quartiles = df['total_spend'].quantile([0.25, 0.5, 0.75])
print(quartiles)

iqr = quartiles[0.75] - quartiles[0.25]
print(f"IQR: £{iqr:.2f}")

# ✅ Q1 £19.99, median £39.98, Q3 £99.95, IQR £79.96
```

> 📝 **Hint:** Code-along Step 2d.

### 💬 What does it mean?

The director asks: *"What does a typical customer spend?"*

**✏️ MODEL ANSWER:**

> A typical transaction is about **£40**. I would report the **median** rather than the mean (£81), because **a small number of very large transactions pull the mean up**: the mean is twice the median, and the largest transaction is £949.95.

## Section 3: Compare groups

```python
# ➡️ Task 3.1: for each product_category, show the count, sum, mean and median of total_spend, rounded to 2 decimal places
category_stats = df.groupby('product_category')['total_spend'].agg(['count', 'sum', 'mean', 'median']).round(2)
category_stats

# ✅ Furniture: 63 transactions, £8,727.99 in total, mean £138.54, median £56.97
```

> 📝 **Hint:** Code-along Step 3a. Save the result as `category_stats` — you will need it in the next task.

```python
# ➡️ Task 3.2: print which category has the highest total revenue, and which has the highest average transaction
print(f"Highest total revenue:     {category_stats['sum'].idxmax()}")
print(f"Highest average transaction: {category_stats['mean'].idxmax()}")

# ✅ Furniture for both
```

> 📝 **Hint:** Code-along Step 3c uses `.idxmax()`.

```python
# ➡️ Task 3.3: calculate each category's share of total revenue, as a percentage rounded to 1 decimal place
category_total = df.groupby('product_category')['total_spend'].sum()
share = (category_total / df['total_spend'].sum() * 100).round(1)
share.sort_values(ascending=False)

# ✅ Furniture 53.6%, Software 23.7%, Electronics 15.6%, Stationery 7.1%
```

> 📝 **Hint:** Code-along Step 3b.

```python
# ➡️ Task 3.4: show the average total_spend for each region, highest first, rounded to 2 decimal places
df.groupby('region')['total_spend'].mean().round(2).sort_values(ascending=False)

# ✅ Yorkshire is highest (£154.53); London is lowest (£56.36)
```

### ➡️ Task 3.5 — Payment methods

`value_counts(normalize=True)` gives each value's **share** (as a decimal) rather than a count. Multiply by 100 for a percentage.

1. Show the percentage of transactions for each `payment_method`, rounded to 1 decimal place.
2. Print the most common payment method using `.mode()[0]`.

```python
# ➡️ Task 3.5: payment method percentages and the most common method
print((df['payment_method'].value_counts(normalize=True) * 100).round(1))
print(f"Most common payment method: {df['payment_method'].mode()[0]}")

# ✅ Credit Card 36.0%, Debit Card 30.0%, PayPal 20.5%, Bank Transfer 8.5%, Voucher 5.0%
```

### 💬 What does it mean?

Which product category should RetailCo prioritise next quarter?

**✏️ MODEL ANSWER:**

> I would prioritise **Furniture**, because it brings in **53.6%** of total revenue and has the highest **average transaction (£138.54)**. Stationery sells often (55 transactions) but brings in only 7.1% of revenue.

## Section 4: Going further with NumPy

```python
# ➡️ Task 4.1: print the standard deviation of total_spend using pandas, and using NumPy
print(f"pandas: £{df['total_spend'].std():.2f}")
print(f"NumPy:  £{np.std(df['total_spend']):.2f}")

# ✅ pandas £118.43, NumPy £118.13 (sample vs population)
```

> 📝 **Hint:** Code-along Step 4b. Why are they different?

### ➡️ Task 4.2 — Outliers with the IQR rule

Use `np.percentile()` to find Q1 and Q3 of `total_spend`, then calculate the **upper fence**: Q3 + 1.5 × IQR. Save it as `upper` and print it.

```python
# ➡️ Task 4.2: calculate and print the upper fence
q1 = np.percentile(df['total_spend'], 25)
q3 = np.percentile(df['total_spend'], 75)
upper = q3 + 1.5 * (q3 - q1)
print(f"Upper fence: £{upper:.2f}")

# ✅ Upper fence: £219.90
```

> 📝 **Hint:** Code-along Step 4c — but this time use the **whole** dataset, not a filtered one.

```python
# ➡️ Task 4.3: use np.where() to create a column called spend_flag ('Outlier' above the fence, 'Normal' otherwise), then count the values
df['spend_flag'] = np.where(df['total_spend'] > upper, 'Outlier', 'Normal')
df['spend_flag'].value_counts()

# ✅ Normal 186, Outlier 14
```

> 📝 **Hint:** Code-along Step 4d.

```python
# ➡️ Task 4.4: which product categories do the outliers come from?
outliers = df[df['spend_flag'] == 'Outlier']
outliers['product_category'].value_counts()

# ✅ Furniture 12, Software 2
```

> 📝 **Hint:** filter to the outliers first (Week 2), then use `value_counts()`.

### ➡️ Task 4.5 — High-value transactions

1. Use `np.where()` to create a column called `value_band`: `'High value'` when `total_spend` is **£100 or more**, and `'Standard'` otherwise.
2. Show the **total revenue** for each value band.

```python
# ➡️ Task 4.5: create value_band, then total revenue for each band
df['value_band'] = np.where(df['total_spend'] >= 100, 'High value', 'Standard')
df.groupby('value_band')['total_spend'].sum().round(2)

# ✅ High value £10,336.67 (from just 42 transactions); Standard £5,946.80
```

### 💬 What does it mean?

Look at your results for Tasks 4.4 and 4.5.

**✏️ MODEL ANSWER:**

> The outliers are mostly **Furniture** transactions (12 of the 14). High-value transactions are only **42** of the 200, but they bring in about **63%** of total revenue (£10,336.67 of £16,283.47). These are genuine large orders, so they should be kept — and they are important customers.

## Section 5: Charts with matplotlib

Remember the pattern: **draw** -> **label** (title and axis labels) -> **show**.

### ➡️ Task 5.3 — A histogram with the mean and median

Histograms are a preview of next week, so this one has blanks to fill in.

> 📝 **Hint:** Code-along Step 5c.

```python
# ➡️ Task 5.3: histogram of total_spend, with the mean and median marked
plt.hist(df['total_spend'], bins=30, edgecolor='white')
plt.axvline(df['total_spend'].mean(), color='red', linestyle='--', label='Mean')
plt.axvline(df['total_spend'].median(), color='black', label='Median')
plt.legend()
plt.title('Transaction values')
plt.xlabel('Total spend (£)')
plt.ylabel('Number of transactions')
plt.show()
```

### 💬 What does it mean?

Look at the histogram in Task 5.3.

> Most transactions are **small (under about £100)**, but a few are **very large (up to £949.95)**. This pulls the mean **well above** the median, which is why the median is the better summary of a typical transaction. (Next week you will learn the name for this shape: right-skewed.)

## Section 6: Limitations

### ➡️ Task 6.1 — One transaction

Remove the **single largest transaction**, then print the mean and median of `total_spend` again. How much does each one change?

> 📝 **Hint:** keep the rows where `total_spend` is less than `df['total_spend'].max()`.

```python
# ➡️ Task 6.1: mean and median without the largest transaction
without_largest = df[df['total_spend'] < df['total_spend'].max()]

print(f"Mean:   with £{df['total_spend'].mean():.2f}, without £{without_largest['total_spend'].mean():.2f}")
print(f"Median: with £{df['total_spend'].median():.2f}, without £{without_largest['total_spend'].median():.2f}")

# ✅ Mean falls from £81.42 to £77.05; the median does not change (£39.98)
```

### ➡️ Task 6.2 — Small groups

`.size()` counts the rows in each group. Group by **both** `region` and `product_category`, then show the **5 smallest** groups.

> 📝 **Hint:** `df.groupby([..., ...]).size().sort_values().head()`

```python
# ➡️ Task 6.2: the 5 smallest region and category groups
df.groupby(['region', 'product_category']).size().sort_values().head()

# ✅ South West / Software has just 1 transaction; several others have only 3 or 4
```

Which **limitation** of descriptive statistics matters most for these results? Choose one and explain why in a sentence:

- **A:** One outlier can pull the mean
- **B:** Two groups can have the same mean but very different spread
- **C:** An overall average can hide differences between groups
- **D:** Small groups give unreliable statistics
- **E:** They describe *what* happened, not *why*

**✏️ MODEL ANSWER:**

A strong answer: **A**, because a few very large orders pull the mean (£81) to twice the median (£40), so the mean gives the wrong idea of a typical customer. **D** is also a good choice: Yorkshire has the highest regional average, but some region and category groups contain only one to four transactions, so their averages could change a lot with one more sale.

## 🔴 Extension 2 — Comparing consistency

The **coefficient of variation** (CV) is the standard deviation divided by the mean. It lets you compare spread between groups with very different averages: the higher the CV, the less consistent the group.

Calculate the CV of `total_spend` for each product category. Which is the least consistent?

```python
# 🔴 Extension 2: coefficient of variation for each category
category_std = df.groupby('product_category')['total_spend'].std()
category_mean = df.groupby('product_category')['total_spend'].mean()
(category_std / category_mean).round(2).sort_values(ascending=False)

# ✅ Furniture is least consistent (1.28); Electronics most consistent (0.71)
```

## 🔴 Extension 3 — The top 10% of transactions

Use `np.percentile()` to find the **90th percentile** of `total_spend`. How many transactions are at or above it, and what share of total revenue do they bring in?

```python
# 🔴 Extension 3: the top 10% of transactions
p90 = np.percentile(df['total_spend'], 90)
top = df[df['total_spend'] >= p90]

print(f"90th percentile: £{p90:.2f}")
print(f"Transactions in the top 10%: {len(top)}")
print(f"Their share of revenue: {top['total_spend'].sum() / df['total_spend'].sum() * 100:.1f}%")

# ✅ £189.99; 22 transactions bring in 47.4% of revenue
```

## 🔴 Extension 4 — A glimpse of SciPy

**SciPy** is another package, with more specialist statistics. A **z-score** says how many standard deviations a value is from the mean; values more than 3 away are often treated as unusual. You will meet these ideas again in Week 5.

```python
# 🔴 Extension 4: z-scores with SciPy
from scipy import stats

df['z_score'] = stats.zscore(df['total_spend'])
df[df['z_score'].abs() > 3][['customer_id', 'product_name', 'total_spend', 'z_score']]

# ✅ 6 transactions have a z-score above 3 (the largest is 7.35).
#    The IQR rule found 14: different methods give different answers, so say which you used.
```
