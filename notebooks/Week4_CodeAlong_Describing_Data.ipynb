## Step 2 — Describe daily revenue

You know these statistics from your statistics module. Here is how to calculate each one in Python using **pandas**:

| Statistic | pandas code |
|-----------|-------------|
| Mean | `df['col'].mean()` |
| Median | `df['col'].median()` |
| Minimum / maximum | `df['col'].min()` / `df['col'].max()` |
| Standard deviation | `df['col'].std()` |
| Quartiles | `df['col'].quantile([0.25, 0.5, 0.75])` |

Try printing each result with an **f-string** (Week 1), so it shows as currency.

````python
# ⌨️ Step 2a: the mean and median daily revenue

````

````python
# ✍️ Step 2b: the smallest, largest and range -> already written: just run this cell
low = df['revenue'].min()
high = df['revenue'].max()

print(f"Lowest day:  £{low:.2f}")
print(f"Highest day: £{high:.2f}")
print(f"Range:       £{high - low:.2f}")

# £150.42 to £1,485.00: a range of £1,334.58
````

````python
# ⌨️ Step 2c: the standard deviation

````

````python
# ✍️ Step 2d: the quartiles and the interquartile range (IQR) -> already written: just run this cell
quartiles = df['revenue'].quantile([0.25, 0.5, 0.75])
print(quartiles)

iqr = quartiles[0.75] - quartiles[0.25]
print(f"IQR: £{iqr:.2f}")

# Q1 £383.68, median £433.44, Q3 £495.50: the middle 50% of days lie within an IQR of £111.81
````

## Step 4 — Going further with NumPy

**NumPy** is the maths package that pandas is built on. For basic statistics, pandas and NumPy usually give the same answer — but not always.

````python
# ✍️ Step 4a: the mean — the same in pandas and NumPy -> already written: just run this cell
print(f"pandas: £{df['revenue'].mean():.2f}")
print(f"NumPy:  £{np.mean(df['revenue']):.2f}")

# Both £446.21
````

### Why are they different?

- **pandas** calculates the **sample** standard deviation (dividing by *n − 1*).
- **NumPy** calculates the **population** standard deviation (dividing by *n*).

With 252 rows the difference is small, but with small groups it can matter. The lesson: **always know what a function is actually calculating.** To make NumPy match pandas, add `ddof=1`:

`np.std(df['revenue'], ddof=1)`

### Outliers — unusual compared with what?

## Step 5 — Creating charts with matplotlib

Every matplotlib chart follows the same pattern:

1. **Draw** the chart, e.g. `plt.bar(x, y)`
2. **Label** it: `plt.title()`, `plt.xlabel()`, `plt.ylabel()`
3. **Show** it: `plt.show()`

Each chart answers **one question**:

| Chart | Question |
|-------|----------|
| Bar chart | How do the groups compare? |
| Boxplot | How spread out is each group, and are there outliers? |
| Histogram | Is the mean a fair summary? (more on this in Week 5) |

## Step 6 — What descriptive statistics can't tell us

Descriptive statistics are powerful, but they have limits:

1. **One outlier can pull the mean** (but not the median).
2. **Two groups can have the same mean but very different spread.** You saw this with University and Station.
3. **An overall average can hide differences between groups.**
4. **Small groups give unreliable statistics.** An average of 3 days is not the same as an average of 84.
5. **They describe *what* happened, not *why*.** They cannot prove causes or predict the future.

Let's see two of these in the data.

````python
# ✍️ Step 6a: limitation 1 — one outlier pulls the mean -> already written: just run this cell
city = df[df['branch'] == 'City Centre']
no_festival = city[city['date'] != '2024-05-18']

print(f"City Centre mean:   with festival £{city['revenue'].mean():.2f}, without £{no_festival['revenue'].mean():.2f}")
print(f"City Centre median: with festival £{city['revenue'].median():.2f}, without £{no_festival['revenue'].median():.2f}")

# One day moves the mean by about £11; the median barely moves (about £1)
````

````python
# ✍️ Step 6b: limitation 3 — the overall average hides the branches -> already written: just run this cell
print("All branches together:")
print(df.groupby('day_type')['revenue'].mean().round(2))
print()
print("Each branch:")
print(df.groupby(['branch', 'day_type'])['revenue'].mean().round(2))

# Overall, weekdays and weekends look almost the same (£449 vs £438).
# But City Centre is much busier at weekends, and University much quieter!
````
