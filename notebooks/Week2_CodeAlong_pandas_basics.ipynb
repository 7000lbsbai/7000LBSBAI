## Step 6 — Group

`df.groupby('branch')['quantity'].sum()` reads as a sentence:

> *"For each branch, take the quantity column and add it up."*

Swap `.sum()` for `.mean()` to get an average, or `.count()` to count rows.

````python
# ⌨️ Step 6a: how many items did each branch sell?
````

````python
# ✍️ Step 6b: average price in each category - already written: just run this cell
df.groupby('category')['price'].mean()

# The same pattern with .mean() instead of .sum()
````

## Step 7 — Create a new column

The data has a price and a quantity, but no order total. We can calculate it for **every row at once** - no loop needed:

`df['new_column'] = calculation`

````python
# ⌨️ Step 7a: create a total column = price × quantity
````

````python
# ✍️ Step 7b: put it together — total revenue for each branch, highest first - already written: just run this cell
df.groupby('branch')['total'].sum().sort_values(ascending=False)

# City Centre £169.80, University £160.80, Station £60.80
````

| Step | Code |
