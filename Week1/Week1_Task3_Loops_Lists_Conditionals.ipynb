## Section 1: Lists — Storing Multiple Values

A **list** is a collection of values stored in a single variable. You create one using square brackets `[]`, with values separated by commas:

```python
prices = [9.99, 24.99, 4.50, 149.00]
```

You can store any data type in a list — numbers, strings, booleans, or even a mix.

### Indexing — accessing individual items

Each item in a list has a numbered position called an **index**. Python counts from **zero**, not one:

```
prices = [9.99, 24.99, 4.50, 149.00]
index:     0      1     2      3
```

To access an item: `prices[0]` returns `9.99`, `prices[3]` returns `149.00`.

Use `len()` to find out how many items are in a list.

## Section 2: For Loops — Processing Every Item Automatically

A **for loop** repeats a block of code once for each item in a list:

```python
for item in my_list:
    # this code runs once for each item
    print(item)
```

On each iteration, `item` takes the value of the next element in the list.

When you need both the **index** (position number) and the **value**, use `enumerate()`:

```python
for index, item in enumerate(my_list):
    print(index, item)
```

This is especially useful when you need to look up matching items from a second list using the same index.

### ✍️ Worked example — printing all transactions

This loop prints every transaction with its customer ID and formatted value.

````python
# ✍️ Worked example — loop over all transactions

print(f"{'Customer':<10} {'Value':>10} {'Category':<15} {'Loyalty':>8}")
print("-" * 47)

for i, value in enumerate(transaction_values):
    customer = customer_ids[i]
    category = categories[i]
    loyalty  = "✓" if loyalty_member[i] else "–"

    print(f"{customer:<10} £{value:>8.2f}  {category:<15} {loyalty:>8}")
````

## Section 3: Combining Loops and Conditionals — Classification

The real power comes from putting a **conditional inside a loop**. This lets you apply business logic to every item automatically:

```python

    else:
        # do something for small items
```

RetailCo classifies transactions into three tiers:

| Transaction value | Tier |
|-------------------|------|

| £50 to £199.99 | 🟡 Medium value |
| Below £50 | 🟢 Low value |

### ✍️ Worked example — classify a single transaction

Before writing the loop, here is how the classification works for one transaction.

````python
# ✍️ Worked example — classify one transaction

value = 19.99

if value >= 200:
    tier = "🔴 High value"
elif value >= 50:
    tier = "🟡 Medium value"
else:
    tier = "🟢 Low value"

print(f"Transaction of £{value:.2f} is classified as: {tier}")
````
