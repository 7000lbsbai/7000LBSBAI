## Section 2: Arithmetic — Calculating Business Metrics

Python supports all the arithmetic you'd expect. Here are the operators we will use most:

| Operator | Symbol | Example | Result |
|----------|--------|---------|--------|
| Addition | `+` | `10 + 3` | `13` |
| Subtraction | `-` | `10 - 3` | `7` |
| Multiplication | `*` | `10 * 3` | `30` |
| Division | `/` | `10 / 3` | `3.333...` |
| Integer division | `//` | `10 // 3` | `3` |
| Remainder | `%` | `10 % 3` | `1` |
| Power | `**` | `2 ** 8` | `256` |

> 📝 **Tip:** Use brackets `()` to control the order of operations — Python follows standard mathematical precedence (BODMAS).

### ✍️ Worked example

Using the Wireless Mouse from Section 1, here is how to calculate total revenue and apply a 20% VAT rate.

````python
# ✍️ Worked example — revenue and VAT calculation

# Re-create the example product
product_name = "Wireless Mouse"
unit_price   = 24.99
quantity     = 150
cost_price   = 12.00    # what RetailCo paid per unit

# Calculate revenue (total income from selling all units)
total_revenue = unit_price * quantity
print("Total revenue: ", total_revenue)

# Calculate total cost and gross profit
total_cost    = cost_price * quantity
gross_profit  = total_revenue - total_cost
print("Total cost:    ", total_cost)
print("Gross profit:  ", gross_profit)

# Calculate profit margin as a percentage
profit_margin = (gross_profit / total_revenue) * 100
print("Profit margin: ", profit_margin)
````

## Section 3: f-strings — Formatted Output

Raw `print()` output is functional but hard to read. **f-strings** let you embed variables directly into text, and control how numbers are displayed.

An f-string starts with the letter `f` before the opening quote:

```python
name  = "Alice"
score = 92.5
print(f"Student: {name} — Score: {score}")
```

You can also control **number formatting** inside the curly braces:

| Format code | What it does | Example |
|-------------|-------------|--------|
| `:.2f` | 2 decimal places | `f"{9.999:.2f}"` → `"10.00"` |
| `:,` | Thousands separator | `f"{12500:,}"` → `"12,500"` |
| `:,.2f` | Both | `f"{12500.5:,.2f}"` → `"12,500.50"` |
| `:.1%` | Percentage, 1 decimal | `f"{0.432:.1%}"` → `"43.2%"` |

### ✍️ Worked example

Here is a well-formatted product summary using f-strings.

````python
# ✍️ Worked example — formatted product summary

product_name  = "Wireless Mouse"
unit_price    = 24.99
quantity      = 150
total_revenue = unit_price * quantity
gross_profit  = total_revenue - (12.00 * quantity)
profit_margin = gross_profit / total_revenue

print("=" * 40)
print(f"  Product:        {product_name}")
print(f"  Unit price:     £{unit_price:.2f}")
print(f"  Units in stock: {quantity:,}")
print(f"  Total revenue:  £{total_revenue:,.2f}")
print(f"  Gross profit:   £{gross_profit:,.2f}")
print(f"  Profit margin:  {profit_margin:.1%}")
print("=" * 40)
````

## Section 4: Conditionals — Business Logic

A **conditional** lets your code make a decision. In Python, we use `if`, `elif` (else if), and `else`:

```python
if condition:          # if this is True...
    # do something
elif other_condition:  # otherwise, if this is True...
    # do something else
else:                  # otherwise...
    # do this
```

**Important:** Python uses **indentation** (4 spaces or a tab) to show what is inside the `if` block. This is not optional — Python will raise an error if your indentation is wrong.

Common **comparison operators**:

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal to | `quantity == 0` |
| `!=` | Not equal to | `in_stock != True` |
| `>` | Greater than | `unit_price > 50` |
| `<` | Less than | `profit_margin < 0.1` |
| `>=` | Greater than or equal | `quantity >= 100` |
| `<=` | Less than or equal | `stock_level <= 10` |

### ✍️ Worked example

RetailCo uses a stock alert system. This code checks the stock level and prints an appropriate message.

````python
# ✍️ Worked example — stock level alert

product_name = "Wireless Mouse"
quantity     = 150

if quantity == 0:
    print(f"{product_name}: OUT OF STOCK — reorder immediately")
elif quantity <= 20:
    print(f"{product_name}: LOW STOCK ({quantity} units) — reorder soon")
elif quantity <= 50:
    print(f"{product_name}: Stock is adequate ({quantity} units)")
else:
    print(f"{product_name}: Stock is healthy ({quantity} units) — no action needed")
````
